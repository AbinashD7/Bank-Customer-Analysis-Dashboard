```python
import pandas as pd
df = pd.read_csv("Banking.csv")
df.head()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df.columns
```

```python
df["client_id"].duplicated().sum()
```

```python
df = df.drop_duplicates(subset=["client_id"], keep="first")
```

```python
df["client_id"].duplicated().sum()
```

```python
df['genderid'] = df['genderid'].replace({1: 'male', 2: 'female'})
```

```python
df.head()
```

```python
df.rename(columns = {
    'joined_bank' : 'joined_date',
    'genderid' : 'gender'
},inplace=True)
```

```python
df.head()
```

```python
lables = ['Young', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels=lables )
```

```python
df[['age', 'age_group']].head(10)
```

```python
df = df.drop(['banking_contact', 'iaid', 'brid'], axis = 1)
```

```python
df = df.drop('location_id', axis = 1)
```

```python
df.head()
```

```python
df['occupation'] = df['occupation'].str.replace(r'\s+(I|II|III|IV)$', '', regex=True)
```

```python
df['occupation'].unique()
```

```python
df["joined_date"].dtype
```

```python
df["joined_date"] = pd.to_datetime(df["joined_date"],format="%d-%m-%Y")
```

```python
df["joined_date"].dtype
```

```python
df["fee_structure"] = pd.qcut(df["estimated_income"],q=3,labels=["Low", "Mid", "High"])
```
